pandas with matplotlib

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df=pd.read_csv("c:/Users/DELL/Documents/pandas/pokemon_data_for_learning/pokemon_gen_1_data.csv")

type_count=df["Type1"].value_counts(ascending=True)

plt.barh(type_count.index,type_count.values,color="cyan",edgecolor="Black")

plt.title("Number of pokemon by primary type in gen1",fontsize="20",fontweight="bold",
            family="Arial")
plt.xlabel("Number of pokemon",color="purple",fontsize="15",family="Arial",fontweight="bold")
plt.ylabel("Typing",color="violet",fontsize="15",fontweight="bold",family="Arial")

plt.tight_layout()
plt.show()

```
