1) Grid lines = helps make plots easier to read by adding reference lines

```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

x=np.array([1,2,3,4,5])
y=np.array([5,10,15,20,25])

plt.plot(x,y,marker=".",markerfacecolor="black",markeredgecolor="black",
             color="cyan",linewidth="2",)

plt.title("Graph for learning",family="arial",fontsize="20",
                               fontweight="bold" )
plt.xlabel("X axis",color="#0f4897",fontsize="20",fontweight="bold")
plt.ylabel("Y axis",color="#2dcd2a",fontsize="20",fontweight="bold")

plt.grid(axis="both",linewidth="2",color="lightgray",linestyle="solid")

plt.show()
```
