1) Figure = The entire canvas
2) Ax = A single plot(subplot)

```python
import matplotlib.pyplot as plt
import numpy as np

figure,axes=plt.subplots(2,2)
x=np.array([1,2,3,4,5])

# first subplot
axes[0,0].plot(x,x*2,color="#891ab5ba")
axes[0,0].set_title("y=2x",color="#564bbaba",fontweight="bold",family="arial")

# second subplot
# axes[0,1].plot(x,x**2,color="#1aabb5bb")
# axes[0,1].set_title("y=x^2",color="#b96dcbf6",fontweight="bold",family="arial")

axes[0,1].bar(x,x**2,color="#1aabb5bb")
axes[0,1].set_title("y=x^2",color="#b96dcbf6",fontweight="bold",family="arial")

# third subplot
axes[1,0].plot(x,x//2,color="lime")
axes[1,0].set_title("y=x//2",color="#78ce7eba",fontweight="bold",fontsize=15,family="arial")

# fourth subplot
axes[1,1].plot(x,-x,color="#ad3d6dba")
axes[1,1].set_title("y=-x",color="#933bceba",fontweight="bold",fontsize=15,family="arial")
plt.tight_layout()

plt.show()


```
