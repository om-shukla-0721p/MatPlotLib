1) Bar charts= compare category of data by representing each category with a bar

```python
import matplotlib.pyplot as plt
import numpy as np

categories=np.array(["Grains","Fruits","Vegetables","Protein","Dairy","Sweets"])
values=np.array([4,2,2,5,2,0])

plt.bar(categories,values,color="#8484ED")
# plt.barh(categories,values,color="#928512")

plt.title("Daily consumption",color="cyan",family="arial",fontweight="bold"
          ,fontsize="25")
plt.xlabel("Food",color="red",family="arial",fontweight="bold",fontsize="18")
plt.ylabel("Quantity",color="red",family="arial",fontweight="bold",fontsize="18")
plt.show()
```
