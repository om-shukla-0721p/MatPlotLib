1) Histograms= A visual representation of the distribution of quantative data.
-              they group values into bins intervals
-              and counts how they fail in each range

```python
import matplotlib.pyplot as plt
import numpy as np

scores=np.random.normal(loc=70,scale=15,size=100)
scores=np.clip(scores,0,100)

plt.hist(scores,bins=15,color="lime",
                edgecolor="black")
plt.title("Exam scores",fontsize=25,color="cyan",
                        fontweight="bold",family="arial")

plt.xlabel("Grades Range",fontsize=15,fontweight="bold",
                                color="#5D49CFB8",family="arial")
plt.ylabel("Number of students",fontsize="15",fontweight="bold",
                                family="arial",color="#9d3ac7ba")

plt.show()

```
