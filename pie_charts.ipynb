1) pie charts= good for visualizing data amoung categories

```python
import numpy as np
import matplotlib.pyplot as plt

categories=["sleeping","coding","mobile","workout","chores","others"]
time_values=[8,4,6,2,1,3]
# colors for differnt categories
colors=["lightgray","cyan","red","#6fbd7d","#525c9a","#c76bd7"]

plt.pie(time_values,labels=categories,
        colors=colors,
        autopct="%1.1f%%",
        explode=[0.1,0,0,0,0,0],
        # explode=[0.1,0.2,0,0,0,0],
        shadow=True,
        startangle=180)
plt.title("Daily routine time chart",color="#398E93",
                                     fontsize="20",
                                     fontweight="bold",
                                     family="arial",
                                     )
plt.show()
```
