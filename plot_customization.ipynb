plot customization


```python
import matplotlib.pyplot as plt
import numpy as np
x=np.array([2023,2024,2025,2026])
y1=np.array([15,25,30,20])
y2=np.array([17,23,38,5])
y3=np.array([13,15,19,20])

line_style=dict(marker=".",markersize=10,markerfacecolor="black",
                        markeredgecolor="red",
                        linestyle="solid",
                        linewidth="2",)
                        # color="#0E8761DC")


# plt.plot(x,y1,marker=".",markersize=10,markerfacecolor="black",
#                         markeredgecolor="red",
#                         linestyle="solid",
#                         linewidth="2",
#                         color="#0E8761DC") 
plt.plot(x,y1,color="#14092BAA",**line_style)
# ms also works as markersize
# mfc is abbreviation for markerfacecolor

plt.plot(x,y2,color="#12768FFF",**line_style)

plt.plot(x,y3,color="#E48181FF",**line_style)
plt.show()

```
