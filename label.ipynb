```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

x=[2020,2021,2022,2023,2024,2025,2026]
y1=np.array([71,73,71,84,90.16,85.4,90])
y2=np.array([95,74,71,82,90.2,84,78.2])
y3=np.array([87,94,91,78,85,63,52])

plt.plot(x,y3,marker=".",markersize=10,
              markerfacecolor="black",
              markeredgecolor="black",
              linewidth=2,
              color="red")

plt.plot(x,y1,color="green",marker="2",
              markersize="10",  
              markerfacecolor="black",
              markeredgecolor="black",
              linewidth=2)

plt.plot(x,y2,color="blue",marker="v",
              markersize="3",markerfacecolor="black",
              markeredgecolor="black",linewidth="2",
              )
# new things
plt.title("Students year wise result",fontsize=14,
          family="Arial",fontweight="bold",
          color="#186479c7")

# label for x axi
plt.xlabel("Year",fontsize="25",family="arial",
           fontweight="bold",color="#1242ab",)

plt.ylabel("marks(in %)",fontsize="25",family="arial",
           fontweight="bold",color="#427a84")

plt.xticks(x)
plt.tick_params(axis="both",
                colors="#13586a")



```
