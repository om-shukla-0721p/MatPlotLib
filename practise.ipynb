```python
import numpy as np
import matplotlib.pyplot as plt

categories=["sleep","mobile","edits","study","socializing","sports","Entertainment","others"]
time_values=np.array([10,2,2,3,1,1,2,3])
# colors for differnt categories
colors=["lightgray","cyan","#6fbd7d","#525c9a","#c76bd7","#a44b57","#d3da4d","#0E0C79"]

plt.pie(time_values,labels=categories,
        colors=colors,
        autopct="%1.1f%%",
        explode=[0,0.1,0.2,0,0,0.1,0,0],
        shadow=True,
        startangle=0)

plt.title("Daily routine chart of Tanisq Mishra",color="#398E93",
           fontsize="18",
           family="Arial",
           fontweight="bold")
plt.plot()
```
