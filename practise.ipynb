```python
import numpy as np
import matplotlib.pyplot as plt

categories=["sleeping/mobile","study","workout","sad","dippresion"]
time_values=np.array([15,3,1,1,4])
# colors for differnt categories
colors=["lightgray","cyan","#6fbd7d","#525c9a","#c76bd7"]

plt.pie(time_values,labels=categories,
        colors=colors,
        autopct="%1.1f%%",
        explode=[0.1,0,0,0,0.1],
        # explode=[0.1,0.2,0,0,0,0],
        shadow=True,
        startangle=360)
plt.title("Daily routine time chart of SAD",color="#398E93",
                                     fontsize="20",
                                     fontweight="bold",
                                     family="arial",
                                     )
plt.show()
```
